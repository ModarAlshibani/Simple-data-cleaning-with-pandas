```python
import pandas as pd
```

```python
dataset = pd.read_csv(r"netflix_titles.csv")
```

```python
dataset
```

# create a new dataset to keep the raw one

```python
new_dataset=dataset.dropna()
```

```python
new_dataset
```

# remove duplicates (if there is any)

```python
new_dataset = dataset.drop_duplicates()
```

# remove the useful columns

here in my database the column "date_added" is not usefull (knowing when the data has been added to the database is not gonna be useful un the future)

```python
new_dataset = new_dataset.drop(columns = "date_added")
new_dataset
```

# Renaming Columns
I want to capitalize the first letter in each column name 

```python
new_columns_names = []
for i in new_dataset.columns:
    new_columns_names.append(i.capitalize())

new_columns_names
```

```python
new_dataset.columns = new_columns_names
new_dataset.head(1)
```

I want to rename the column Listed_in and make it "Genre"

```python
new_dataset = new_dataset.rename(columns={'Listed_in':'Genre'})
```

```python
new_dataset.head(1)
```

# Removing the 's' from the id column

```python
new_dataset['Show_id'].apply(lambda x: x.split("s")[1])
```

```python
new_dataset['Show_id'] = new_dataset['Show_id'].apply(lambda x: x.split("s")[1])
```

```python
new_dataset
```

the show_id was a string because of its old form ... so now I'm gonna change the type of the id to int

# Change Show_id to int

```python
#before ..
type(new_dataset["Show_id"][1])
```

```python
new_dataset["Show_id"] = new_dataset["Show_id"].astype(int)
```

```python
#now ...
type(new_dataset["Show_id"][1])
```

```python
test = new_dataset.dropna()
```

```python
test
```

# Delaing with NaN values

here in this dataset .... if one of the duration or the Country or the Rating is missing for the show then i don't want it 


```python
new_dataset.dropna(subset=['Release_year', 'Country', 'Rating'], inplace=True)
```

```python
new_dataset
```

# Filling the missing values with 'Unknown'

```python
new_dataset['Director'].fillna('Unknown', inplace=True)
new_dataset['Cast'].fillna('Unknown', inplace=True)
```

```python
new_dataset
```

# Rewriting the Show_id after deleting the rows

```python
new_dataset = new_dataset.reset_index(drop=True)
```

```python
new_dataset
```

```python
new_dataset['Show_id'] = range(0, len(new_dataset))
```

```python
new_dataset
```

```python
new_dataset.to_csv("cleaned_netflix_titles.csv")
```
